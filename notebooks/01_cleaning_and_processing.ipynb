Substituição das vírgulas por pontos

```python
df = df.apply( lambda x: x.str.replace(',', '.'))
```

Transformação das variáveis categóricas(str) para númericas(float) e substituição dos traços (---) pela média

```python
df['Janeiro'] = pd.to_numeric(df["Janeiro"].str.replace('---', '305.79'))
df['Fevereiro'] = pd.to_numeric(df['Fevereiro'].str.replace('---', '249.3'))
df['Março'] = pd.to_numeric(df['Março'].str.replace('---', '227.97'))
df['Abril'] = pd.to_numeric(df['Abril'].str.replace('---', '93.64'))
df['Maio'] = pd.to_numeric(df['Maio'].str.replace('---', '86.67'))
df['Junho'] = pd.to_numeric(df['Junho'].str.replace('---', '59.66'))
df['Julho'] = pd.to_numeric(df['Julho'].str.replace('---', '54.6'))
df['Agosto'] = pd.to_numeric(df['Agosto'].str.replace('---', '27.85'))
df['Setembro'] = pd.to_numeric(df['Setembro'].str.replace('---', '83.8'))
df['Outubro'] = pd.to_numeric(df['Outubro'].str.replace('---', '139.89'))
df['Novembro'] = pd.to_numeric(df['Novembro'].str.replace('---', '145.04'))
df['Dezembro'] = pd.to_numeric(df['Dezembro'].str.replace('---', '236.04'))

```
